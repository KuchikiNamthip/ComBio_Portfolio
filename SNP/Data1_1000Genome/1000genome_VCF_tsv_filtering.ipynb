# Step 2 Clear and prep data

## 1. Keep only SNPs (and split multi-allelic sites)

```python
import pandas as pd

# 5.0 Ensure canonical columns exist
base_cols = ['CHROM','POS','ID','REF','ALT','QUAL','FILTER','INFO']
missing = [c for c in base_cols if c not in df.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

# 5.1 Split multi-allelic records (ALT = 'A,C') into separate rows
df = df.assign(ALT=df['ALT'].str.split(',')).explode('ALT')

# 5.2 Enforce SNPs only (single-base REF/ALT; drop anything else)
is_snp = (df['REF'].str.len() == 1) & (df['ALT'].str.len() == 1)
df = df[is_snp].copy()

df.info()
```

```python
# 5.3 Basic de-dup + sorting
df['POS'] = pd.to_numeric(df['POS'], errors='coerce')
df = df.dropna(subset=['POS'])
df = df.drop_duplicates(subset=['CHROM','POS','REF','ALT'])
df = df.sort_values(['CHROM','POS'])

df.info()

print(f"✅ After SNP-only & split: {len(df):,} rows")
df[['CHROM','POS','REF','ALT','QUAL','FILTER']].head()
```

## 2. Basic QC (FILTER & QUAL)

```python
# Accept PASS or '.' (some callers leave '.')
pass_mask = df['FILTER'].isin(['PASS','.'])
df = df[pass_mask].copy()

# Optional QUAL ≥ 30 (only if present/meaningful for your caller)
df['QUAL_num'] = pd.to_numeric(df['QUAL'], errors='coerce')
df = df[(df['QUAL'].eq('.')) | (df['QUAL_num'] >= 30)].drop(columns=['QUAL_num'])

print(f"✅ After FILTER/QUAL: {len(df):,} rows")
df.info()
```

```python
import pandas as pd

# --- Parse each INFO field into a dict ---
def parse_info(s):
    info_dict = {}
    if pd.isna(s):
        return info_dict
    for kv in str(s).split(';'):
        if '=' in kv:
            k, v = kv.split('=', 1)
            info_dict[k] = v
        elif kv:  # flag (no value)
            info_dict[k] = True
    return info_dict

# Convert INFO strings to dict objects
df['INFO_dict'] = df['INFO'].map(parse_info)

# Inspect one
print(df['INFO_dict'].iloc[0])
```

## 1) Coerce numerics & derive handy metrics

## 3) Define thresholds (interview-friendly defaults)

These are typical, reasonable defaults that you can explain:

- Rarity: global MAF < 1% (or 0.5% if you want stricter)

- Across populations: make sure it’s not common in any subpopulation

- Quality: MQ ≥ 40, QD ≥ 2, VQSLOD > 0 (when present)

- Depth: total site depth not too low (avoid ultra-low coverage calls). For 1000G sites, use a soft floor (e.g., DP ≥ 100)

- Completeness: NS close to cohort size (avoid sites genotyped in very few samples)

```python
# ---- Rarity thresholds ----
GLOBAL_MAF_MAX = 0.01   # keep variants with MAF < 1% globally
POP_MAF_MAX    = 0.01   # and < 1% in every subpopulation present

# ---- Quality thresholds (used only if present in INFO) ----
MQ_MIN, QD_MIN, VQSLOD_MIN = 40, 2, 0
DP_MIN, NS_MIN = 100, 1500  # site-level depth and #samples with data

# Keys for population AFs that may exist in your INFO
POP_KEYS = ['EAS_AF','EUR_AF','AFR_AF','AMR_AF','SAS_AF']
```

## 4) Helpers to read from INFO_dict safely

```python
import pandas as pd

def fget(d, key, cast=float, default=None):
    """Safe getter from INFO_dict with casting and comma-list handling."""
    if not isinstance(d, dict): return default
    v = d.get(key, default)
    if v is None: return default
    if isinstance(v, str) and ',' in v:
        v = v.split(',', 1)[0]
    try:
        return cast(v) if cast else v
    except Exception:
        return default
```

## 5) Build the single filtering function

```python
def pass_info_filter(d):
    # 1) SNP-only if VT is present
    vt = fget(d, 'VT', str, '')
    if vt and vt != 'SNP':
        return False

    # 2) Global rarity (use AF or derive from AC/AN)
    af = fget(d, 'AF', float, None)
    if af is None:
        ac, an = fget(d, 'AC', float, None), fget(d, 'AN', float, None)
        af = (ac / an) if (ac is not None and an not in (None, 0)) else None
    maf = (af if (af is not None and af <= 0.5) else (1 - af) if af is not None else None)
    if maf is not None and maf >= GLOBAL_MAF_MAX:
        return False  # common globally

    # 3) Rare in every subpopulation that exists in this record
    for k in POP_KEYS:
        v = fget(d, k, float, None)
        if v is not None and v >= POP_MAF_MAX:
            return False

    # 4) Quality (skip check if field is absent)
    dp  = fget(d, 'DP', float, None)
    ns  = fget(d, 'NS', float, None)
    mq  = fget(d, 'MQ', float, None)
    qd  = fget(d, 'QD', float, None)
    vqs = fget(d, 'VQSLOD', float, None)

    if dp  is not None and dp  < DP_MIN:        return False
    if ns  is not None and ns  < NS_MIN:        return False
    if mq  is not None and mq  < MQ_MIN:        return False
    if qd  is not None and qd  < QD_MIN:        return False
    if vqs is not None and vqs <= VQSLOD_MIN:   return False

    return True
```

## 6) Apply the filter and inspect results

```python
# Respect site FILTER too (if present at top level, not inside INFO)
if 'FILTER' in df.columns:
    df = df[df['FILTER'].isin(['PASS','.'])].copy()

mask = df['INFO_dict'].map(pass_info_filter)
candidates = df.loc[mask, ['CHROM','POS','ID','REF','ALT','FILTER','INFO', 'MAF']].copy()

# (Optional) surface a few INFO values for quick viewing
candidates['AF']      = df.loc[mask, 'INFO_dict'].map(lambda d: fget(d, 'AF', float, None))
candidates['DP']      = df.loc[mask, 'INFO_dict'].map(lambda d: fget(d, 'DP', float, None))
candidates['NS']      = df.loc[mask, 'INFO_dict'].map(lambda d: fget(d, 'NS', float, None))
for k in POP_KEYS:
    candidates[k] = df.loc[mask, 'INFO_dict'].map(lambda d, k=k: fget(d, k, float, None))

print(f"✅ Candidates after INFO-only filters: {len(candidates):,}")
candidates.head(10)
```

## 7) Save a tidy TSV

```python
candidates.sort_values(['CHROM','POS']).to_csv(
    "candidates_from_INFO_dict.tsv",
    sep='\t', index=False
)
print("Wrote: candidates_from_INFO_dict.tsv")
```

```python
from pyensembl import EnsemblRelease

ensembl = EnsemblRelease(105)  # GRCh38
def annotate(chrom, pos):
    genes = ensembl.genes_at_locus(contig=chrom, position=pos)
    return ", ".join(g.gene_name for g in genes) if genes else None

df['gene'] = df.apply(lambda x: annotate(x['CHROM'], x['POS']), axis=1)
```

➡️ Then add effect info from VEP’s REST API:

```python
import requests
def vep_annot(chrom, pos, ref, alt):
    url = f"https://rest.ensembl.org/vep/human/region/{chrom}:{pos}/{ref}/{alt}?"
    r = requests.get(url, headers={"Content-Type": "application/json"})
    if r.ok:
        data = r.json()[0]
        return data.get("most_severe_consequence")
    return None

df['effect'] = df.apply(lambda x: vep_annot(x['CHROM'], x['POS'], x['REF'], x['ALT']), axis=1)
```
